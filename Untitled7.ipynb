# Qwen2.5-VL Server (Colab)

Ce notebook lance un serveur FastAPI expose publiquement (ngrok) pour Qwen seulement.
Le serveur recoit une image base64 + un prompt et retourne une reponse textuelle.

```python
import base64
import io
import threading
from dataclasses import dataclass

import nest_asyncio
import torch
import uvicorn
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from PIL import Image
from pyngrok import ngrok
from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration

nest_asyncio.apply()
```

```python
@dataclass
class ServerConfig:
    model_id: str = "Qwen/Qwen2.5-VL-3B-Instruct"
    host: str = "0.0.0.0"
    port: int = 8000
    max_new_tokens: int = 180

cfg = ServerConfig()

device_map = "auto"
torch_dtype = torch.bfloat16 if torch.cuda.is_available() else torch.float32

model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    cfg.model_id,
    torch_dtype=torch_dtype,
    device_map=device_map,
)
processor = AutoProcessor.from_pretrained(cfg.model_id)
```

```python
class AnalyzeRequest(BaseModel):
    prompt: str
    image_base64: str
    max_new_tokens: int = 180
    temperature: float = 0.2

app = FastAPI(title="Qwen2.5-VL Server")

@app.get("/health")
def health():
    return {"ok": True, "model": cfg.model_id}

@app.post("/qwen/analyze")
def analyze(req: AnalyzeRequest):
    try:
        image_bytes = base64.b64decode(req.image_base64)
        image = Image.open(io.BytesIO(image_bytes)).convert("RGB")
    except Exception as exc:
        raise HTTPException(status_code=400, detail=f"Image invalide: {exc}")

    messages = [
        {
            "role": "user",
            "content": [
                {"type": "image", "image": image},
                {"type": "text", "text": req.prompt},
            ],
        }
    ]

    text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = processor(text=[text], images=[image], return_tensors="pt")
    inputs = {k: v.to(model.device) if hasattr(v, "to") else v for k, v in inputs.items()}

    generated = model.generate(
        **inputs,
        max_new_tokens=min(max(16, int(req.max_new_tokens)), 512),
        temperature=float(req.temperature),
        do_sample=float(req.temperature) > 0,
    )
    output_ids = generated[:, inputs["input_ids"].shape[-1]:]
    answer = processor.batch_decode(output_ids, skip_special_tokens=True)[0].strip()
    return {"answer": answer}
```

```python
# Option 1: mettre ton token ngrok en variable d'environnement COLAB_NGROK_TOKEN
# import os
# ngrok.set_auth_token(os.getenv("COLAB_NGROK_TOKEN"))

public_url = ngrok.connect(cfg.port).public_url
print("Qwen public URL:", public_url)
print("Health endpoint:", f"{public_url}/health")
print("Analyze endpoint:", f"{public_url}/qwen/analyze")
```

```python
def run_server():
    uvicorn.run(app, host=cfg.host, port=cfg.port, log_level="info")

thread = threading.Thread(target=run_server, daemon=True)
thread.start()
print("Serveur Qwen lance.")
```
